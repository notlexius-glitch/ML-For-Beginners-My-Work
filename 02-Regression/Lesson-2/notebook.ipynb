# Regression Lesson 2: Data Preparation and Visualization

This notebook explores the US Pumpkin dataset, prepares the data, and creates visualizations using Matplotlib and Seaborn.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pumpkins = pd.read_csv('../../2-Regression/data/US-pumpkins.csv')

pumpkins.head()
```

```python
columns_to_select = ['Package', 'Low Price', 'High Price', 'Date']

pumpkins = pumpkins.loc[:, columns_to_select]

pumpkins.head()
```

```python
pumpkins = pumpkins[pumpkins['Package'].str.contains('bushel', case=True, regex=True)]

pumpkins.head()
```

```python
price = (pumpkins['Low Price'] + pumpkins['High Price']) / 2

month = pd.DatetimeIndex(pumpkins['Date']).month

new_pumpkins = pd.DataFrame({
    'Month': month,
    'Package': pumpkins['Package'],
    'Low Price': pumpkins['Low Price'],
    'High Price': pumpkins['High Price'],
    'Price': price
})

new_pumpkins.head()
```

```python
new_pumpkins.groupby(['Month'])['Price'].mean().plot(kind='bar')

plt.ylabel("Average Pumpkin Price")
plt.title("Average Pumpkin Price by Month")

plt.show()
```

```python
sns.catplot(
    x="Month",
    y="Price",
    data=new_pumpkins,
    kind="bar"
)

plt.show()
```

```python
correlations = new_pumpkins[['Month', 'Low Price', 'High Price', 'Price']].corr()

sns.heatmap(correlations, annot=True)

plt.show()
```

## Conclusion

Matplotlib helped create customizable graphs, while Seaborn made statistical visualization easier with less code.

The visualizations helped identify pumpkin price patterns and relationships between features.
