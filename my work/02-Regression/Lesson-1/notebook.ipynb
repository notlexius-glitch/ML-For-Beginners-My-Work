# Regression with Scikit-learn

This notebook explores the Linnerud dataset and builds regression models to understand relationships between exercise and physiological measurements.


```python
import matplotlib.pyplot as plt
from sklearn import datasets, linear_model
from sklearn.model_selection import train_test_split
```

```python
data = datasets.load_linnerud()

print(data.feature_names)
print(data.target_names)
```

```python
X = data.data[:, 0].reshape(-1, 1)
y = data.target[:, 0]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = linear_model.LinearRegression()
model.fit(X_train, y_train)

prediction = model.predict(X_test)
```

```python
plt.scatter(X_test, y_test)
plt.plot(X_test, prediction)

plt.xlabel("Waistline")
plt.ylabel("Situps")
plt.title("Waistline vs Situps Regression")

plt.show()
```

```python
X = data.data[:, 2].reshape(-1, 1)
y = data.target[:, 1]

model.fit(X, y)

plt.scatter(X, y)
plt.plot(X, model.predict(X))

plt.xlabel("Weight")
plt.ylabel("Chinups")
plt.title("Weight vs Chinups Regression")

plt.show()
```
